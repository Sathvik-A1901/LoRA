## Step 3: Fine-Tune the Model with Unsloth

````python
import torch
from unsloth import FastLanguageModel
from trl import SFTTrainer
from transformers import TrainingArguments
from datasets import load_dataset

# ==== Configuration ====
model_name = "unsloth/llama-3-8b-Instruct-bnb-4bit" # Recommended model
data_path = "ollama_training.jsonl"
output_dir = "unsloth_finetuned_adapters"

# ==== Load Model with Unsloth ====
max_seq_length = 2048
dtype = None
load_in_4bit = True

model, tokenizer = FastLanguageModel.from_pretrained(
    model_name = model_name,
    max_seq_length = max_seq_length,
    dtype = dtype,
    load_in_4bit = load_in_4bit,
)

# ==== Add LoRA Adapters ====
model = FastLanguageModel.get_peft_model(
    model,
    r = 16,
    target_modules = ["q_proj", "k_proj", "v_proj", "o_proj",
                      "gate_proj", "up_proj", "down_proj"],
    lora_alpha = 16,
    lora_dropout = 0,
    bias = "none",
    use_gradient_checkpointing = True,
    random_state = 3407,
)

# ==== Load and Format Dataset ====
# We create a new 'text' column that formats the prompt and completion
# for instruction fine-tuning.
instruction_template = "### Instruction:\
{instruction}\
\n### Response:\
{response}"

def formatting_prompts_func(examples):
    instructions = examples["prompt"]
    responses = examples["completion"]
    texts = []
    for instruction, response in zip(instructions, responses):
        text = instruction_template.format(instruction=instruction, response=response)
        texts.append(text)
    return { "text" : texts, }
pass

dataset = load_dataset("json", data_files=data_path, split = "train")
dataset = dataset.map(formatting_prompts_func, batched = True,)

# ==== Configure Trainer ====
trainer = SFTTrainer(
    model = model,
    tokenizer = tokenizer,
    train_dataset = dataset,
    dataset_text_field = "text",
    max_seq_length = max_seq_length,
    dataset_num_proc = 2,
    packing = False,
    args = TrainingArguments(
        per_device_train_batch_size = 2,
        gradient_accumulation_steps = 4,
        warmup_steps = 5,
        max_steps = 60, # You can increase this for longer training
        learning_rate = 2e-4,
        fp16 = not torch.cuda.is_bf16_supported(),
        bf16 = torch.cuda.is_bf16_supported(),
        logging_steps = 1,
        optim = "adamw_8bit",
        weight_decay = 0.01,
        lr_scheduler_type = "linear",
        seed = 3407,
        output_dir = output_dir,
    ),
)

# ==== Start Training ====
print("Starting training...")
trainer.train()
print("Training complete!")
````

## Step 4: Save Model to GGUF for Ollama

Unsloth makes it easy to save the fine-tuned model in the GGUF format that Ollama uses. This cell will merge the LoRA adapters and quantize the model to 4-bit precision.

````python
# Save to GGUF
from google.colab import files

gguf_filename = "finetuned_llama3_8b.gguf"
print(f"Saving to GGUF format: {gguf_filename}")

model.save_pretrained_gguf(gguf_filename, tokenizer, quantization_method = "q4_k_m")

print(f"Model saved. Now downloading {gguf_filename}...")
files.download(gguf_filename)
````

## Step 5: Run Your Model in Ollama

Once the `finetuned_llama3_8b.gguf` file is downloaded, you can run it in Ollama.

1.  **Move the GGUF file:** Place the downloaded `.gguf` file in a directory of your choice.

2.  **Create a `Modelfile`:** In the same directory, create a new file named `Modelfile` (with no extension) and add the following content:

    ```
    FROM ./finetuned_llama3_8b.gguf
    TEMPLATE """### Instruction:
    {{ .Prompt }}
    
    ### Response:
    """
    ```

3.  **Create the Ollama Model:** Open your terminal and run the following command from that directory:

    ```bash
    ollama create resume-crafter -f Modelfile
    ```

4.  **Run the Model:** You can now chat with your personalized model!

    ```bash
    ollama run resume-crafter
    ```
